```python
# ============================================================
# STEP 26: THRESHOLD ANALYSIS
# ============================================================

import numpy as np
import pandas as pd

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

def threshold_analysis(y_true, probabilities, model_name):

    thresholds = np.arange(0.05, 0.96, 0.05)

    results = []

    for threshold in thresholds:

        predictions = (
            probabilities >= threshold
        ).astype(int)

        tn, fp, fn, tp = confusion_matrix(
            y_true,
            predictions,
            labels=[0, 1]
        ).ravel()

        precision = precision_score(
            y_true,
            predictions,
            zero_division=0
        )

        recall = recall_score(
            y_true,
            predictions,
            zero_division=0
        )

        f1 = f1_score(
            y_true,
            predictions,
            zero_division=0
        )

        false_alarm_rate = fp / (fp + tn)

        results.append({
            "threshold": threshold,
            "TP": tp,
            "FP": fp,
            "FN": fn,
            "TN": tn,
            "precision": precision,
            "recall": recall,
            "f1": f1,
            "false_alarm_rate": false_alarm_rate
        })

    results_df = pd.DataFrame(results)

    print("\n" + "=" * 80)
    print(model_name)
    print("=" * 80)

    display(
        results_df.round(3)
    )

    print("\nBest threshold by F1:")

    best = results_df.loc[
        results_df["f1"].idxmax()
    ]

    display(
        best.to_frame().T.round(3)
    )

    return results_df


# ------------------------------------------------------------
# Logistic Regression
# ------------------------------------------------------------

lr_thresholds = threshold_analysis(
    y_test,
    y_test_prob,
    "LOGISTIC REGRESSION — TEST"
)


# ------------------------------------------------------------
# Random Forest
# ------------------------------------------------------------

rf_thresholds = threshold_analysis(
    y_test,
    rf_test_prob,
    "RANDOM FOREST — TEST"
)
```

```python
# ============================================================
# STEP 28 — FINAL TEST EVALUATION
# LOCKED VALIDATION THRESHOLD
# ============================================================

from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    roc_auc_score,
    average_precision_score,
    precision_score,
    recall_score,
    f1_score
)

# ------------------------------------------------------------
# LOCKED THRESHOLD
# ------------------------------------------------------------

FINAL_THRESHOLD = 0.75

y_test_pred_final = (
    y_test_prob >= FINAL_THRESHOLD
).astype(int)


# ------------------------------------------------------------
# METRICS
# ------------------------------------------------------------

tn, fp, fn, tp = confusion_matrix(
    y_test,
    y_test_pred_final,
    labels=[0, 1]
).ravel()

precision = precision_score(
    y_test,
    y_test_pred_final,
    zero_division=0
)

recall = recall_score(
    y_test,
    y_test_pred_final,
    zero_division=0
)

f1 = f1_score(
    y_test,
    y_test_pred_final,
    zero_division=0
)

far = fp / (fp + tn)

roc_auc = roc_auc_score(
    y_test,
    y_test_prob
)

pr_auc = average_precision_score(
    y_test,
    y_test_prob
)


# ------------------------------------------------------------
# PRINT
# ------------------------------------------------------------

print("=" * 70)
print("FINAL LOGISTIC REGRESSION — TEST")
print("=" * 70)

print(f"Locked threshold: {FINAL_THRESHOLD}")

print(f"\nROC-AUC: {roc_auc:.4f}")
print(f"PR-AUC:  {pr_auc:.4f}")

print("\nConfusion Matrix:")
print(
    confusion_matrix(
        y_test,
        y_test_pred_final
    )
)

print("\nTP:", tp)
print("FP:", fp)
print("FN:", fn)
print("TN:", tn)

print(f"\nPrecision: {precision:.4f}")
print(f"Recall:    {recall:.4f}")
print(f"F1:        {f1:.4f}")
print(f"False Alarm Rate: {far:.4f}")

print("\nClassification Report:")
print(
    classification_report(
        y_test,
        y_test_pred_final,
        digits=4,
        zero_division=0
    )
)
```

```python
def extract_era5_point(ds, row):

    point = ds.sel(
        time=np.datetime64(row["datetime_utc"]),
        latitude=row["latitude"],
        longitude=row["longitude"],
        method="nearest"
    )

    result = {
        "storm_id": row["storm_id"],
        "datetime_utc": row["datetime_utc"],
        "latitude": row["latitude"],
        "longitude": row["longitude"],
        "RI_24h": row["RI_24h"],

        # 850 hPa
        "vo_850": float(
            point["vo"].sel(isobaricInhPa=850).values
        ),
        "d_850": float(
            point["d"].sel(isobaricInhPa=850).values
        ),
        "u_850": float(
            point["u"].sel(isobaricInhPa=850).values
        ),
        "v_850": float(
            point["v"].sel(isobaricInhPa=850).values
        ),
        "rh_850": float(
            point["r"].sel(isobaricInhPa=850).values
        ),
        "t_850": float(
            point["t"].sel(isobaricInhPa=850).values
        ),

        # 700 hPa
        "vo_700": float(
            point["vo"].sel(isobaricInhPa=700).values
        ),
        "d_700": float(
            point["d"].sel(isobaricInhPa=700).values
        ),
        "rh_700": float(
            point["r"].sel(isobaricInhPa=700).values
        ),
        "t_700": float(
            point["t"].sel(isobaricInhPa=700).values
        ),

        # 500 hPa
        "vo_500": float(
            point["vo"].sel(isobaricInhPa=500).values
        ),
        "d_500": float(
            point["d"].sel(isobaricInhPa=500).values
        ),
        "rh_500": float(
            point["r"].sel(isobaricInhPa=500).values
        ),
        "t_500": float(
            point["t"].sel(isobaricInhPa=500).values
        ),

        # 200 hPa
        "vo_200": float(
            point["vo"].sel(isobaricInhPa=200).values
        ),
        "d_200": float(
            point["d"].sel(isobaricInhPa=200).values
        ),
        "u_200": float(
            point["u"].sel(isobaricInhPa=200).values
        ),
        "v_200": float(
            point["v"].sel(isobaricInhPa=200).values
        ),
    }

    return result
```

```python
# ================================================================
# ERA5 → IMD FEATURE EXTRACTION — MVP
# ================================================================

import os
import glob
import numpy as np
import pandas as pd
import xarray as xr

print("=" * 70)
print("ERA5 → IMD FEATURE EXTRACTION")
print("=" * 70)

# ------------------------------------------------
# 1. Load IMD-ERA5 observation table
# ------------------------------------------------

imd_files = glob.glob("/content/*IMD*ERA5*.csv")

print("\nPossible IMD-ERA5 files:")
for f in imd_files:
    print(os.path.basename(f))

# Prefer the ready file if it exists
ready_files = [
    f for f in imd_files
    if "ready" in os.path.basename(f).lower()
]

if ready_files:
    imd_file = ready_files[0]
else:
    imd_file = imd_files[0]

print("\nUsing:")
print(imd_file)

df = pd.read_csv(imd_file)

print("\nIMD rows:", len(df))
print("Columns:")
print(df.columns.tolist())


# ------------------------------------------------
# 2. Identify required columns
# ------------------------------------------------

# Rename common possibilities into standard names

column_map = {}

for col in df.columns:
    c = col.lower()

    if c in ["storm_id", "stormid", "sid"]:
        column_map[col] = "storm_id"

    elif "datetime" in c and "utc" in c:
        column_map[col] = "datetime_utc"

    elif c == "datetime":
        column_map[col] = "datetime_utc"

    elif c in ["latitude", "lat"]:
        column_map[col] = "latitude"

    elif c in ["longitude", "lon", "lng"]:
        column_map[col] = "longitude"

    elif "ri_24h" in c:
        column_map[col] = "RI_24h"

df = df.rename(columns=column_map)

required = [
    "storm_id",
    "datetime_utc",
    "latitude",
    "longitude",
    "RI_24h"
]

missing = [c for c in required if c not in df.columns]

if missing:
    raise ValueError(
        f"Missing required columns: {missing}\n"
        f"Available columns: {df.columns.tolist()}"
    )

df["datetime_utc"] = pd.to_datetime(df["datetime_utc"])

print("\nRequired columns found ✓")


# ------------------------------------------------
# 3. ERA5 files
# ------------------------------------------------

era5_files = sorted(
    glob.glob("/content/ERA5_chunks_fixed/*.grib")
)

print("\nERA5 files:", len(era5_files))


# Create date → ERA5 file lookup
era5_lookup = {}

for f in era5_files:

    name = os.path.basename(f)

    # ERA5_YYYY-MM-DD.grib
    date_string = name.replace(
        "ERA5_", ""
    ).replace(
        ".grib", ""
    )

    try:
        date = pd.Timestamp(date_string).date()
        era5_lookup[date] = f
    except:
        pass

print("ERA5 dates indexed:", len(era5_lookup))


# ------------------------------------------------
# 4. Variables and pressure levels
# ------------------------------------------------

levels = [850, 700, 500, 200]

variables = ["d", "r", "t", "u", "v"]


# ------------------------------------------------
# 5. Extract
# ------------------------------------------------

results = []

successful = 0
missing_dates = 0
failed = 0

for i, row in df.iterrows():

    dt = row["datetime_utc"]

    lat = float(row["latitude"])
    lon = float(row["longitude"])

    date = dt.date()

    print(
        f"\rProcessing {i+1}/{len(df)}",
        end=""
    )

    # Find ERA5 file for this date
    era_file = era5_lookup.get(date)

    if era_file is None:
        missing_dates += 1
        continue

    try:

        # Open GRIB
        ds = xr.open_dataset(
            era_file,
            engine="cfgrib",
            backend_kwargs={
                "indexpath": ""
            }
        )

        # Find nearest ERA5 time
        time_values = pd.to_datetime(
            ds.time.values
        )

        time_diff = np.abs(
            time_values - np.datetime64(dt)
        )

        nearest_idx = int(
            np.argmin(time_diff)
        )

        nearest_time = time_values[
            nearest_idx
        ]

        delta_minutes = abs(
            (nearest_time - dt).total_seconds()
        ) / 60.0

        # Only accept reasonably close ERA5 time
        if delta_minutes > 180:
            ds.close()
            failed += 1
            continue

        # Extract nearest grid point
        point = ds.sel(
            time=nearest_time,
            latitude=lat,
            longitude=lon,
            method="nearest"
        )

        feature_row = {
            "storm_id": row["storm_id"],
            "datetime_utc": dt,
            "latitude": lat,
            "longitude": lon,
            "RI_24h": row["RI_24h"],
            "era5_datetime": nearest_time,
            "era5_delta_minutes": delta_minutes
        }

        # Extract all variables at all pressure levels
        for var in variables:

            for level in levels:

                name = f"{var}_{level}"

                value = point[var].sel(
                    isobaricInhPa=level
                ).values

                feature_row[name] = float(
                    np.asarray(value)
                )

        # ------------------------------------------------
        # Derived vertical wind shear
        # ------------------------------------------------

        u850 = feature_row["u_850"]
        v850 = feature_row["v_850"]

        u200 = feature_row["u_200"]
        v200 = feature_row["v_200"]

        feature_row["shear_850_200"] = np.sqrt(
            (u200 - u850)**2 +
            (v200 - v850)**2
        )

        results.append(feature_row)

        successful += 1

        ds.close()

    except Exception as e:

        failed += 1

        try:
            ds.close()
        except:
            pass


print("\n\n" + "=" * 70)
print("ERA5 EXTRACTION COMPLETE")
print("=" * 70)

print("IMD observations :", len(df))
print("Successful        :", successful)
print("Missing dates     :", missing_dates)
print("Failed            :", failed)


# ------------------------------------------------
# 6. Save
# ------------------------------------------------

era5_df = pd.DataFrame(results)

output_file = "/content/RI_ERA5_features_MVP.csv"

era5_df.to_csv(
    output_file,
    index=False
)

print("\nSaved:")
print(output_file)

print("\nShape:")
print(era5_df.shape)

print("\nColumns:")
print(era5_df.columns.tolist())

print("\nFirst rows:")
display(era5_df.head())
```

```python
# ================================================================
# ERA5-ONLY XGBOOST — MVP
# ================================================================

import pandas as pd
import numpy as np

from sklearn.model_selection import GroupShuffleSplit
from sklearn.metrics import (
    roc_auc_score,
    average_precision_score,
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)
from xgboost import XGBClassifier

print("=" * 70)
print("ERA5-ONLY RI MODEL")
print("=" * 70)

# ------------------------------------------------
# Load ERA5 features
# ------------------------------------------------

df = pd.read_csv("/content/RI_ERA5_features_MVP.csv")

df["datetime_utc"] = pd.to_datetime(df["datetime_utc"])

print("Rows:", len(df))
print("Storms:", df["storm_id"].nunique())

print("\nRI distribution:")
print(df["RI_24h"].value_counts())


# ------------------------------------------------
# Features
# ------------------------------------------------

feature_cols = [
    "d_850", "d_700", "d_500", "d_200",
    "r_850", "r_700", "r_500", "r_200",
    "t_850", "t_700", "t_500", "t_200",
    "u_850", "u_700", "u_500", "u_200",
    "v_850", "v_700", "v_500", "v_200",
    "shear_850_200"
]

X = df[feature_cols].copy()
y = df["RI_24h"].astype(int)
groups = df["storm_id"]


# ------------------------------------------------
# Remove bad values
# ------------------------------------------------

X = X.replace([np.inf, -np.inf], np.nan)

valid = X.notna().all(axis=1)

X = X.loc[valid]
y = y.loc[valid]
groups = groups.loc[valid]
df_model = df.loc[valid].copy()

print("\nUsable rows:", len(X))


# ------------------------------------------------
# STORM-SAFE SPLIT
# ------------------------------------------------

gss1 = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    gss1.split(X, y, groups)
)

X_train = X.iloc[train_idx]
X_test = X.iloc[test_idx]

y_train = y.iloc[train_idx]
y_test = y.iloc[test_idx]

groups_train = groups.iloc[train_idx]
groups_test = groups.iloc[test_idx]

print("\n" + "=" * 70)
print("STORM-SAFE SPLIT")
print("=" * 70)

print("Train:", len(X_train))
print("Test :", len(X_test))

print("Train storms:", groups_train.nunique())
print("Test storms :", groups_test.nunique())

print("\nTrain RI:")
print(y_train.value_counts())

print("\nTest RI:")
print(y_test.value_counts())


# ------------------------------------------------
# CLASS WEIGHT
# ------------------------------------------------

n0 = (y_train == 0).sum()
n1 = (y_train == 1).sum()

scale_pos_weight = n0 / n1

print("\nscale_pos_weight:", scale_pos_weight)


# ------------------------------------------------
# XGBOOST
# ------------------------------------------------

model = XGBClassifier(
    n_estimators=300,
    max_depth=3,
    learning_rate=0.03,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="binary:logistic",
    eval_metric="logloss",
    scale_pos_weight=scale_pos_weight,
    random_state=42
)

print("\n" + "=" * 70)
print("TRAINING ERA5 XGBOOST")
print("=" * 70)

model.fit(
    X_train,
    y_train
)

print("✅ TRAINING COMPLETE")


# ------------------------------------------------
# PREDICTIONS
# ------------------------------------------------

p_test = model.predict_proba(X_test)[:, 1]

# Standard threshold
pred_test = (p_test >= 0.5).astype(int)


# ------------------------------------------------
# METRICS
# ------------------------------------------------

roc = roc_auc_score(y_test, p_test)
pr = average_precision_score(y_test, p_test)

precision = precision_score(
    y_test,
    pred_test,
    zero_division=0
)

recall = recall_score(
    y_test,
    pred_test,
    zero_division=0
)

f1 = f1_score(
    y_test,
    pred_test,
    zero_division=0
)

cm = confusion_matrix(
    y_test,
    pred_test
)

print("\n" + "=" * 70)
print("ERA5-ONLY TEST RESULTS")
print("=" * 70)

print(f"ROC-AUC : {roc:.4f}")
print(f"PR-AUC  : {pr:.4f}")
print(f"Precision: {precision:.4f}")
print(f"Recall   : {recall:.4f}")
print(f"F1       : {f1:.4f}")

print("\nConfusion Matrix:")
print(cm)


# ------------------------------------------------
# SAVE MODEL
# ------------------------------------------------

model.save_model(
    "/content/era5_only_xgboost_mvp.json"
)

print("\nModel saved:")
print("/content/era5_only_xgboost_mvp.json")


# ------------------------------------------------
# SAVE PREDICTIONS
# ------------------------------------------------

test_results = df_model.iloc[test_idx].copy()

test_results["P_RI"] = p_test
test_results["prediction"] = pred_test

test_results.to_csv(
    "/content/ERA5_only_test_predictions.csv",
    index=False
)

print("\nPredictions saved:")
print("/content/ERA5_only_test_predictions.csv")
```
